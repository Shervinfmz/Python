## ♻️  Itertools - Efficient Looping and Combinations

The built-in **`itertools`** module is a powerful library for creating efficient, memory-saving iterators that produce complex sequences for specific looping patterns. These functions are often faster than manually creating the same lists or sequences.

### The Code Example

This code demonstrates various functions for infinite cycling, counting, and generating mathematical combinations and permutations:

```python
import itertools as it

# --- 1. Infinite Iterators: cycle() ---
print("--- Infinite Iterator: cycle() ---")
a = it.cycle([1, 2, 3])
print(f"Next(a): {next(a)}") # Output: 1
print(f"Next(a): {next(a)}") # Output: 2
print(f"Next(a): {next(a)}") # Output: 3
print(f"Next(a) (Cycle Restart): {next(a)}") # Output: 1
print(f"Next(a): {next(a)}") # Output: 2
print("-" * 30)

# --- 2. Infinite Iterators: count() ---
print("--- Infinite Iterator: count() ---")
# count(start, step)
counter = it.count(3, 10)
print(f"Counter: {counter}") # The iterator object itself
print(f"Next(counter): {next(counter)}") # Start: 3
print(f"Next(counter): {next(counter)}") # 3 + 10 = 13
print(f"Next(counter): {next(counter)}") # 13 + 10 = 23
print("-" * 30)


# --- 3. Combinatorics: Permutations and Combinations ---
numbers = [1, 2, 3, 4]

# Combinations: Order does NOT matter.
print("--- Combinations ---")
combinations_result = list(it.combinations(numbers, 2))
print(f"combinations(nums, 2): {combinations_result}") # Output: [(1, 2), (1, 3), (1, 4), (2, 3), (2, 4), (3, 4)]

# Combinations with Replacement: Includes repeats.
combinations_wr_result = list(it.combinations_with_replacement(numbers, 2))
print(f"combinations_with_replacement(nums, 2): {combinations_wr_result}")

# Permutations: Order DOES matter.
print("--- Permutations ---")
permutations_result = list(it.permutations(numbers[:3], 2)) # Using [1, 2, 3]
print(f"permutations(nums[:3], 2): {permutations_result}") # Output: [(1, 2), (2, 1), (1, 3), (3, 1), (2, 3), (3, 2)]
print("-" * 30)

# --- 4. Functional Tools: accumulate() ---
print("--- Functional Tool: accumulate() ---")
# Default operation is addition (running sum)
accumulate_result = list(it.accumulate(numbers))
print(f"accumulate(nums): {accumulate_result}") # Output: [1, 3, 6, 10]
```

### Explanation and Key Concepts

The `itertools` module focuses on **iterators**, which produce values one at a time and save memory, unlike functions that return entire lists at once.

#### 1. Infinite Iterators (`cycle`, `count`)

These functions create iterators that theoretically run forever, so you must use `next()` or `break` when looping to stop them.

* **`it.cycle(iterable)`**: Repeats the elements of an iterable indefinitely. It is perfect for circular queues or repeating tasks.
* **`it.count(start, step)`**: Creates an iterator that returns evenly spaced values starting with `start`. If `step` is omitted, it defaults to 1. This is often used as a simple infinite counter.

#### 2. Combinatorial Iterators (`combinations`, `permutations`)

These functions are used in statistics, data science, and algorithm development to find all possible groupings of items. 

| Function | What it Generates | Order Matters? | Replacement? |
| :--- | :--- | :--- | :--- |
| **`combinations(iterable, r)`** | Unique sequences of length `r`. | **No** (e.g., (1, 2) is the same as (2, 1)) | No |
| **`combinations_with_replacement(r)`**| Unique sequences of length `r`, including repeats. | No | **Yes** |
| **`permutations(iterable, r)`** | All ordered sequences of length `r`. | **Yes** (e.g., (1, 2) is different from (2, 1)) | No |

#### 3. Functional Tools (`accumulate`)

* **`it.accumulate(iterable, func=operator.add)`**: Returns an iterator that yields accumulated results. By default, it performs a **running sum** (cumulative addition), but you can pass an optional second function (e.g., multiplication from the `operator` module) to perform other cumulative operations.
