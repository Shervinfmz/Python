## 📝  Simple OOP - Class Definition and Instantiation

This tutorial focuses on the most basic components of Object-Oriented Programming (OOP): defining a Class (the blueprint) and creating an Instance (the object) that holds its own data via a Constructor.

### The Code Example

This code demonstrates querying types, finding element positions, and creating a basic class with a constructor:

```python
# --- 1. Basic List Operations (Review from Tutorial 4) ---
my_list = [4, 3, 5]
# Use type() to identify the object's class
print(f"Type of my_list: {type(my_list)}") 
# Find the index (position) of the first item with value 3
print(f"Index of value 3: {my_list.index(3)}") 

# --- 2. String Operation (Review from Tutorial 5) ---
name = "shervin"
# Counts the number of times the substring 'shervin' appears within the string 'shervin' (result is 1)
print(f"Count of '{name}' within '{name}': {name.count(name)}") 

# --- 3. Simple Class Definition ---
class Test(): # CLASS: The blueprint for creating Test objects
    pass # 'pass' means the class has no attributes or methods yet

test_instance = Test() # INSTANCE: Creating a concrete object from the class blueprint
# Check the type of the instance
print(f"Type of test_instance: {type(test_instance)}")

# --- 4. Class with Constructor (__init__) ---
class Book():
    # CONSTRUCTOR: This method runs automatically when a new object is created
    def __init__(self, page_count): 
        # ATTRIBUTE: Attaching the input value (page_count) to the instance itself (self.pages)
        self.pages = page_count 

mybook = Book(540) # INSTANTIATION: Creates the object, passing 540 to the constructor

# ACCESS: Retrieve the unique attribute data from the object
print(f"Book Pages Attribute: {mybook.pages}")
# Check the final type
print(f"Type of mybook: {type(mybook)}")
```

#### 1. Defining a Class

* **`class Test():`**: This defines a **Class**. A class is like a custom data type; it serves as a template for objects. The `pass` keyword is used here because the class has no attributes or methods yet.
* **`test_instance = Test()`**: This creates an **Instance (Object)** named `test_instance` from the `Test` blueprint.
* **`type(test_instance)`**: This confirms that the object's type is the class it was created from.

#### 2. The Constructor and Attributes

The `Book` class demonstrates how an object gets its unique data. 

* **`class Book():`**: Defines the blueprint for all book objects.
* **`def __init__(self, page_count):`**: This is the **Constructor**. It's automatically called when `Book(540)` is executed.
* **`self`**: Refers to the object being built (`mybook`).
* **`page_count`**: The input value (`540`) provided during instantiation.
* **`self.pages = page_count`**: This creates a specific **Attribute** (`pages`) on the `mybook` instance and sets its value to `540`.
* **`mybook.pages`**: This accesses the unique `pages` attribute stored inside the `mybook` object.
