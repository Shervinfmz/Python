## 🧑‍💻 Object-Oriented Programming (OOP) Basics

Object-Oriented Programming (OOP) is a paradigm that structures programs around objects, which bundle data (attributes) and methods (functions) together. This tutorial covers the core mechanics of creating a class and using it to generate instances (objects).

### The Code Example

This code defines a simple class, creates an instance, and demonstrates accessing its properties and calling its method:

```python
class ClassName():
    # 1. The Constructor Method (Initializer)
    def __init__(self, param1): # attribute
        # 'self' refers to the instance being created
        self.param1 = param1
        print("object created")

    # 2. A Custom Method
    def say_hello(self): # method
        print(f"Hello from instance with param1={self.param1}")

# 3. Creating an Instance (Object)
t = ClassName(5) # The value 5 is passed as param1 to __init__

# 4. Accessing the Attribute
print(f"Attribute value: {t.param1}")

# 5. Calling the Method
t.say_hello()

# 6. Checking the Type
print(f"Type of object t: {type(t)}")
```

### Explanation and Key Concepts

#### 1. The Class (`class ClassName():`)

A class is a blueprint or a template for creating objects. It defines the structure (attributes) and behavior (methods) that all objects of that type will possess. 

#### 2. The Constructor (`def __init__(self, ...):`)

* **Purpose:** The `__init__` method is automatically called every time a new instance (object) of the class is created. It initializes the object's attributes.
  
* **`self`:** This is the most crucial keyword in Python classes. It is the first argument of every instance method and refers to the **instance itself**. When you write `self.param1 = param1`, you are attaching the value of the input parameter (`param1`) to the instance's attribute (`self.param1`).
  
* **Attributes:** Variables attached to an instance (like `t.param1`) are called attributes. They hold the data specific to that object.

#### 3. Methods (`def say_hello(self):`)

* **Definition:** Methods are functions defined inside a class. They describe the actions or behaviors the object can perform.
  
* **Instance Access:** Methods always take `self` as their first argument, allowing them to access and modify the instance's attributes (e.g., accessing `self.param1` inside `say_hello`).

#### 4. The Instance/Object (`t = ClassName(5)`)

* **Instantiation:** This process creates a concrete object (`t`) from the class blueprint (`ClassName`).
  
* **When `ClassName(5)` is called:**
  
    * A new object (`t`) is created.
    * The `__init__(self, param1)` method is called, passing the new object as `self` and `5` as `param1`.
    * The object's attribute `t.param1` is set to `5`.
  
* **Type:** The type of the object `t` is the class it was instantiated from (`<class '__main__.ClassName'>`).
