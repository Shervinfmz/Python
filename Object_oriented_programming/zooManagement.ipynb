A zoo wants to create an animal management system. You are required to design this system using **Object-Oriented Programming (OOP)** in Python.

## Tasks:

### 1. **Introduction to OOP**  
Your task is to define a class called `Animal` with the following attributes:

- `name`: name of the animal
- `species`: species of the animal
- `age`: age of the animal
- `sound`: sound the animal makes

Then, create an instance of this class for a **"Lion"** and print its details.

### 2. **Attributes and Class Methods**  
Add a method `make_sound()` to the `Animal` class that prints the sound of the animal.

### 3. **Class Object Attributes and Methods**  
Add a class attribute `zoo_name` that holds the name of the zoo. Also, create a method `info()` that prints the details of the animal.

### 4. **Inheritance and Polymorphism**  
Create a subclass called `Bird` that inherits from `Animal`. The `Bird` class should have a new attribute `wing_span` (for the bird's wingspan).  
Override the `make_sound()` method in the `Bird` class to make the sound specific to birds.

### 5. **Magic Methods**  
Implement the `__str__` method in the `Animal` class so that when an object is printed, it displays the details of the animal.




```python
class Animal():
    def __init__(self,name, species, age, sound):
        zoo_name="city zoo"

        self.name= name
        self.species= species
        self.age=age
        self.sound=sound

    def make_sound(self):
        print(f"the {self.species} make {self.sound} sound!")
    
    def info(self):
        print(f"This is a {self.age} years old {self.species} named {self.name}.")
    def __str__(self):
        return f"{self.name} is a {self.age} years old {self.species} makes a {self.sound} sound."

class Bird(Animal):
    def __init__(self, name, species, age, sound,wing_span):
        super().__init__(name, species, age, sound)
        self.wing_span=wing_span

    def make_sound(self):
        print(f"the {self.species} makes a chriping sound!")

    def info(self):
        super().info()
        print(f"It has a wingspan of {self.wing_span} meters!")

l= Animal("lion","lion",5, "roar")
print(l)
l.make_sound()
l.info()
sparrow=Bird("sparrow","Bird",2, "chrip", 0.25)
sparrow.info()
```
