## 💻 Python Coding Challenge: OOP, Inheritance, and Polymorphism

### 1. Create the Base Class: $\text{Vehicle}$

Define a base class named $\text{Vehicle}$ with the following requirements:

Attributes ($\text{\_\_init\_\_}$): $\text{make}$, $\text{model}$, $\text{year}$, and $\text{color}$.

Instance Method ($\text{description}$): A method that prints a brief description of the vehicle, including all its attributes.

### 2. Create Subclasses: $\text{Car}$ and $\text{Motorcycle}$

Create two subclasses that inherit from $\text{Vehicle}$:

Subclass $\text{Car}$:

Add a new attribute: $\text{fuel\_type}$.

Use $\text{super()}$ in the constructor to initialize the parent class attributes.

Override the $\text{description()}$ method to include the $\text{fuel\_type}$ information.

Subclass $\text{Motorcycle}$:

Add a new attribute: $\text{engine\_type}$.

Use $\text{super()}$ in the constructor to initialize the parent class attributes.

Override the $\text{description()}$ method to include the $\text{engine\_type}$ information.

### 3. Demonstrate Polymorphism

Write the execution part of the code:

Create an instance of $\text{Vehicle}$ ($\text{vehicle1}$).

Create an instance of $\text{Car}$ ($\text{car1}$).

Create an instance of $\text{Motorcycle}$ ($\text{motorcycle1}$).

Create a list named $\text{vehicles}$ containing these three objects.

Iterate through the $\text{vehicles}$ list and call the $\text{description()}$ method for each object in the loop.

This loop will demonstrate polymorphism, as the same method call ($\text{v.description()}$) will execute different versions of the code based on the object's actual type.

```python
class Vehicle():
    def __init__(self, producer, model, year, color):
        self.producer = producer
        self.model = model
        self.year = year
        self.color = color

    def description(self):
        print(f"The producer of this vehicle is {self.producer}, "
              f"which has the {self.model} model, produced in {self.year}, and its color is {self.color}.")

class Car(Vehicle):
    def __init__(self, producer, model, year, color, fuel_type):
        super().__init__(producer, model, year, color)
        self.fuel_type = fuel_type

    def description(self):
        print(f"The producer of this vehicle is {self.producer}, "
              f"which has the {self.model} model, produced in {self.year}, color is {self.color}, "
              f"and it uses {self.fuel_type} fuel.")

class Motorcycle(Vehicle):
    def __init__(self, producer, model, year, color, engine_type):
        super().__init__(producer, model, year, color)
        self.engine_type = engine_type

    def description(self):
        print(f"The producer of this vehicle is {self.producer}, "
              f"which has the {self.model} model, produced in {self.year}, color is {self.color}, "
              f"and it has a {self.engine_type} engine.")

  
vehicle1 = Vehicle("Toyota", "Corolla", 2020, "Red")
vehicle1.description()

car1 = Car("Honda", "Civic", 2021, "Blue", "Petrol")
car1.description()

motorcycle1 = Motorcycle("Yamaha", "YZF-R15", 2022, "Black", "4-stroke")
motorcycle1.description()

vehicles=[vehicle1,car1,motorcycle1]
for v in vehicles:
    v.description()
```
