```python
class Book:
    Book_type = "horror"

    def __init__(self, name, page):
        self.pages = page
        self.name = name

    def open(self):
        print(f"Opened the {self.name} which has {self.pages} pages")


b1 = Book("C programming", 224)
b1.open()


class Lesson (Book): #inheritance 
    def __init__(self, major, grade, name, pages):
        Book.__init__(self,name, pages)
        print("a new lesson in book")
        self.major= major
        self.grade= grade
    def open(self):
        print(f"opened {self.name} of {self.major} major {self.grade}")

l=Lesson ("math",3, "programming guide", 500)
type(l)
l.pages
l.name
l.major
l.open()

class Runner:
    def __init__(self, name):   #iinheritance
        self.name = name

    def action(self):
        print(f"{self.name} is running")

# استفاده
sara = Runner("sara")
sara.action()


class cyclist:
    def __init__(self, name):   #iinheritance
        self.name = name

    def action(self):
        print(f"{self.name} is biking")

shervin= cyclist("shervin faranarzi")#polymorphism
shervin.action()
for person in (shervin, sara):
    person.action
def show_what_doing (person):
    person.action()

show_what_doing(shervin)
show_what_doing(sara)

class Human():
    def __init__(self,name):
        pass
    def jump(self):
        raise NotImplementedError ("implement the jump")
    

class programer(Human):
    pass
    def jump(self):
       print("I jumped")

j=programer("shervin")
j.jump()
```
