## 💎 Advanced OOP - Inheritance, Polymorphism, Abstraction

This tutorial covers the advanced pillars of Object-Orientened Programming (OOP): **Inheritance** (reusing code), **Polymorphism** (using one interface for multiple types), and **Abstraction** (defining required behavior).

### The Code Example

This code block demonstrates how a subclass (`Lesson`) inherits from a superclass (`Book`) and how method overriding and polymorphism work across different classes (`Runner`, `cyclist`, `Human`):

```python
# --- Superclass Definition ---
class Book:
    # Class Attribute (Shared by all instances)
    book_type = "horror" 

    def __init__(self, name, page):
        # Instance Attributes (Unique to each object)
        self.pages = page
        self.name = name

    def open(self):
        """Method to open the book."""
        print(f"Opened the {self.name} which has {self.pages} pages")


# --- Inheritance Example (Lesson inherits from Book) ---
class Lesson(Book): 
    def __init__(self, major, grade, name, pages):
        # 1. OPTIMIZATION/FIX: Use super() to call the parent's constructor. 
        # This correctly initializes self.name and self.pages.
        super().__init__(name, pages) 
        print("A new lesson object created.")
        
        # 2. Adding new attributes specific to Lesson
        self.major = major
        self.grade = grade
        
    def open(self):
        """OVERRIDING: Replaces Book's open() method for Lesson objects."""
        print(f"Opened {self.name} of {self.major} major (Grade {self.grade}).")

# Creation of instances
l = Lesson("Math", 3, "Programming Guide", 500)
l.open()
print("-" * 30)


# --- Polymorphism Example ---

class Runner:
    def __init__(self, name): 
        self.name = name

    def action(self):
        print(f"{self.name} is running.")

class Cyclist:
    def __init__(self, name):
        self.name = name

    def action(self): # Shares the interface (method name) with Runner
        print(f"{self.name} is biking.")

sara = Runner("Sara")
shervin = Cyclist("Shervin Faramarzi")

# Demonstrate Polymorphism via a loop
print("--- Polymorphism Demonstration ---")
for person in (shervin, sara):
    person.action() 

# --- Abstraction Example ---

class Human():
    def __init__(self, name):
        # CRITICAL FIX: The name attribute MUST be initialized here 
        # to prevent AttributeError in subclasses.
        self.name = name 

    def jump(self):
        """Enforces implementation in subclasses (Abstraction)."""
        raise NotImplementedError("Subclasses must implement the 'jump' method.")


class Programmer(Human): # Programmer is a concrete implementation of Human
    # No need for __init__ if it only uses the parent's arguments
    def jump(self):
        """Concrete implementation of the required method."""
        print(f"Programmer {self.name} executed the jump logic.")

j = Programmer("Shervin")
j.jump()
```

### Explanation of OOP Pillars

#### 1. Inheritance (Code Reusability)

* **Concept:** Inheritance allows a **child class** (`Lesson`) to inherit attributes and methods from a **parent class** (`Book`). 


* **Method Overriding:** The `Lesson.open()` method replaces the parent's `Book.open()` method for all `Lesson` instances.
* **Parent Constructor Call:** In `Lesson.__init__`, the line `Book.__init__(self, name, pages)` is necessary to initialize the attributes defined in the `Book` class (`name` and `pages`).

#### 2. Polymorphism (One Interface, Many Forms)

* **Concept:** Polymorphism means "many forms." It allows objects of different classes (e.g., `Runner` and `Cyclist`) to be processed using a single common interface (the `action()` method).
* **Demonstration:** The `for` loop calls `person.action()` on both a `Runner` and a `Cyclist`. The correct, specific behavior is executed each time without needing an `if/else` check for the object type.

#### 3. Abstraction (Defining Behavior)

* **Concept:** Abstraction defines the necessary behavior for objects without providing the implementation details. 

* **Enforcing Methods:** The `Human.jump()` method uses `NotImplementedError` to establish a contract: any concrete subclass (like `Programmer`) inheriting from `Human` *must* provide its own working `jump` implementation.
* **Concrete Implementation:** The `Programmer` class fulfills this contract, ensuring the method exists and works.
