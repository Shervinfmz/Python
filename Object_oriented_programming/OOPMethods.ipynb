## 🧠 Class vs. Instance Attributes and Methods

This tutorial covers the distinction between **Class Attributes** (data shared by all instances), **Instance Attributes** (data unique to each object), and **Methods** (the actions an object can perform). Understanding these components is fundamental to OOP in Python.

### The Code Example

This code demonstrates how different elements (attributes and methods) are defined, accessed, and modified:

```python
class Book():
    # 1. CLASS ATTRIBUTE (Shared by all objects)
    book_type = "horror" 

    def __init__(self, page_count):
        # 2. INSTANCE ATTRIBUTE (Unique to each object)
        self.pages = page_count 

    # 3. INSTANCE METHOD (Action the object can perform)
    def open(self):
        print(f"Opening the book on last page ({self.pages})")


# --- Demonstration of Instance Attributes (pages) ---
b = Book(400)
print("--- Instance Attributes (Pages) ---")
print(f"b.pages: {b.pages}")
b.open() # Calls the method
print("-" * 30)


# --- Demonstration of Class Attributes (book_type) ---
b1 = Book(340)
b2 = Book(500)
print("--- Shared Class Attribute Access ---")
print(f"b1.book_type (Default): {b1.book_type}") 
print(f"b2.book_type (Default): {b2.book_type}") 
print("-" * 30)

# --- Modification: Instance Override vs. Class Change ---

# 4. INSTANCE OVERRIDE: Modifying b1 creates a NEW instance attribute
b1.book_type = "funny"
print("--- Modification Effects ---")
print(f"b1.book_type (Override): {b1.book_type}") 
print(f"b2.book_type (Still Default): {b2.book_type}") 

# 5. CLASS CHANGE: Modifying the class attribute changes the default for ALL future/unmodified instances
Book.book_type = "triller"
b3 = Book(10) # New instance created AFTER the class change

print(f"b3.book_type (New Default): {b3.book_type}") 
print(f"b2.book_type (Still Uses New Default): {b2.book_type}")
print(f"b1.book_type (Still Uses Override): {b1.book_type}") 
print("-" * 30)

# --- Practical Example: Circle Area ---
class Circle():
    pi = 3.1415926 # Class Attribute

    def __init__(self, r):
        self.r = r # Instance Attribute

    def area(self):
        # Accessing instance data (r) and class data (pi)
        m = self.r * self.r * Circle.pi 
        return m

c1 = Circle(10) 
print("--- Circle Area Example ---")
print(f"Calculated Area: {c1.area()}")
print("-" * 30)
```

### Explanation and Key Concepts

#### 1. Methods (`def open(self):`)

Methods define the behavior and actions an object can perform.

* **Definition:** A function defined inside the class.
* **Instance Access:** Methods always receive `self` as the first parameter, allowing them to access the object's unique data (e.g., `self.pages` inside the `open` method).

#### 2. Instance Attributes (Unique Data)

* **Definition:** Defined inside the `__init__` constructor using the `self` keyword (`self.pages`).
* **Scope:** Unique to each object (instance).
* **Access:** Accessed using the instance variable (`b.pages` or `c1.r`).

#### 3. Class Attributes (Shared Data)

* **Definition:** Defined directly inside the class body but outside any method (`book_type`, `pi`).
* **Scope:** Shared by **all** instances of that class. They act as constants or default values for the entire class.
* **Access:** Accessed either through the instance (`b1.book_type`) or, preferably, through the class itself (`Book.book_type`).

#### 4. The Crucial Difference in Modification

* **Modifying the INSTANCE (`b1.book_type = "funny"`):** Creates a new instance attribute on `b1` only, **overriding** the class default for that specific object.
* **Modifying the CLASS (`Book.book_type = "triller"`):** Changes the default value for all instances that have **not** been individually overridden.
