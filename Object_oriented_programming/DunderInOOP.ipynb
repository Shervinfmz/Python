## 🧙‍♂️  Magic Methods (`__len__`, `__str__`, `__del__`)

**Magic Methods** (or Dunder Methods) are special methods in Python classes that start and end with double underscores (e.g., `__init__`, `__str__`). They allow you to define how your objects behave with built-in functions, operators, and language features, making your custom classes feel like native Python types.

### The Code Example

This code demonstrates how three specific magic methods are implemented to customize the `Book` object's behavior:

```python
class Book():
    def __init__(self, name, page):
        self.pages = page
        self.name = name

    def open(self):
        print(f"Opened the {self.name} which has {self.pages} pages")
    
    # 1. Magic Method: Enables the built-in len() function
    def __len__(self):
        """Called by len(object). Returns the length (number of pages)."""
        return self.pages
        
    # 2. Magic Method: Enables string representation for print()
    def __str__(self):
        """Called by print(object) or str(object). Returns a user-friendly string."""
        r = f"Book Title: {self.name}, Total Pages: {self.pages}"
        return r
        
    # 3. Magic Method: The Destructor
    def __del__(self):
        """Called when the object is about to be destroyed (deleted from memory)."""
        print(f"Oh, the {self.name} book is vanishing! (Object destroyed)")


# --- Demonstration ---
b = Book("C programming", 242)

print("--- 1. Using __str__ (via print()) ---")
print(b) # Automatically calls b.__str__()

print("\n--- 2. Using __len__ (via len()) ---")
print(f"The length (pages) of b is: {len(b)}") # Automatically calls b.__len__()

print("\n--- 3. Using __del__ (via del keyword) ---")
# Manually deleting the object reference calls __del__ immediately
del b 
# Note: __del__ is also called automatically when the object leaves scope (e.g., at the end of a script)
```

### Explanation and Key Concepts

#### 1. `__str__(self)`: String Representation

* **Purpose:** This method defines the "informal" or user-friendly string representation of an object.
* **Trigger:** It is automatically called when you use the built-in functions `print(b)` or `str(b)`.
* **Requirement:** It **must** return a string. Without `__str__`, printing the object would only show its memory address (`<__main__.Book object at 0x...>`). 

#### 2. `__len__(self)`: Length Operation

* **Purpose:** This method defines what the "length" of your custom object should be.
* **Trigger:** It is automatically called when you use the built-in function `len(b)`.
* **Requirement:** It **must** return an integer (or raise an appropriate exception if length is not applicable).

#### 3. `__del__(self)`: The Destructor

* **Purpose:** This method is called the **destructor**. It is executed when the object's reference count drops to zero—meaning the object is no longer being used by any part of the program and is about to be destroyed by the Python garbage collector.
* **Trigger:** It is often called explicitly by the `del` keyword (`del b`), or automatically when the script ends or the object goes out of scope.
