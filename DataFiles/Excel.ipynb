## 📘 Working With Excel Files in Python Using openpyxl

This tutorial explains how to work with Excel files in Python using the openpyxl library.

### ✅ 1. Install openpyxl

To install the library, run

python -m pip install openpyxl


### ✅ 2. Load an Excel Workbook

```python
import openpyxl

# Load an existing Excel file
wb = openpyxl.load_workbook("FinancialSample.xlsx")

# Check the type
type(wb)
```

### ✅ 3. Access the Active Worksheet

```python
ws = wb.active
print(ws)
```

### ✅ 4. Read Data From Cells
Read the cell object:

```python
print(ws["A4"])
```

Read the actual value:

```python
this_cell = ws["F2"]
print(this_cell.value)
```

### ✅ 5. Create a New Excel Workbook

```python
from openpyxl import Workbook

new_wb = Workbook()
ws = new_wb.active
```

### ✅ 6. Write Data to Cells

```python
ws["A1"] = "Shervin"        # Write text
ws["F5"] = 4                # Write number
ws.append([3, 2, 4, 5, 6])  # Append an entire row
```

### ✅ 7. Insert Current Date and Time

```python
import datetime

ws["A2"] = datetime.datetime.now()
```

### ✅ 8. Save the Workbook

```python
new_wb.save("funny.xlsx")
```

### ✅ 9. Sum Values in Column F

The loop below sums values from column F (rows 2 to 10), skipping empty cells:

```python
t = 0
for i in range(2, 11):
    value = ws[f"F{i}"].value
    if value is not None:
        t += value

print(t)
```

## 📌 Summary

Use openpyxl.load_workbook() to load an Excel file

Use Workbook() to create a new one

Access or modify cells using ws["A1"] syntax

Add entire rows using append()

Write timestamps with datetime

Save your changes with save()

Loop through cells to calculate totals
