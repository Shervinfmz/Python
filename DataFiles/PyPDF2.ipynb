# 📄 Working With PDFs in Python Using PyPDF2

This tutorial explains how to open a PDF, count its pages, and extract text using the `PyPDF2` library.

---

## ✅ 1. Install PyPDF2

```bash
python -m pip install PyPDF2


## ✅ 2. Import PyPDF2 and Load a PDF File

````python
import PyPDF2

# Load PDF file
pdf = PyPDF2.PdfReader("sample-local-pdf.pdf")
````

## ✅ 3. Inspect the PDF Object

````python
dir(pdf)
````

## ✅ 4. Get Number of Pages

````python
print(len(pdf.pages))
````

## ✅ 5. Access the First Page

````python
first_page = pdf.pages[0]
print(first_page)
````

## ✅ 6. Extract Text From a Page

````python
text = first_page.extract_text()
print(text)
````

## 📌 Summary of PyPDF2 Usage

| Function / Command | Purpose |
|--------------------|---------|
| `PyPDF2.PdfReader("file.pdf")` | Loads a PDF file into Python for reading |
| `len(pdf.pages)` | Returns the total number of pages in the PDF |
| `pdf.pages[i]` | Retrieves page *i* (zero-based index) from the PDF |
| `page.extract_text()` | Extracts text content from the specified page |
| `dir(pdf)` | Shows available attributes and methods of the PDF object |
| `PdfReader.pages` | A list-like object containing all pages |
| `PdfReader.metadata` | Provides access to PDF metadata (title, author, etc.) |

This summary provides the essential commands needed to navigate and extract information from PDF files using **PyPDF2**.
