## 🚀 Flask Tutorial: Building Your First Web App (Hello World)

This tutorial introduces Flask, a lightweight Python web framework, and guides you through setting up a basic web server that responds with a "Hello World" message.

Prerequisites

You need Python installed, and you must install the Flask library:

pip install Flask

**The Code Example:** app.py

This code defines a minimal web application.

```python
from flask import Flask

# 1. Create the Flask Application Instance
app = Flask(__name__)

# 2. Define a Route (URL)
@app.route('/')
def home():
    # 3. Define the Function to Execute when the Route is accessed
    return 'Hello World'

# 4. Run the Application
if __name__ == '__main__':
    app.run(debug=True)
```

### Explanation and Key Concepts

This simple program contains the four essential components of a basic Flask application:

+   from flask import Flask:This imports the Flask class, which is the core component of the framework.

+    **`app = Flask(__name__)`**

     *   This creates an instance of the Flask class, making it your application object.

     *   `__name__` is a special Python variable that holds the name of the current module. Flask uses this to know where to look for templates, static files, and other resources.

+    **@app.route('/'):**

     *   This is a decorator—a special function that modifies another function.

      *  The decorator tells Flask: "When a user navigates to the root URL (/) of the application, execute the function immediately below this line." This process is called routing.

+    **def home(): return 'Hello World':**

      *  This function is the view function (or handler). When the route (/) is accessed, this function runs and returns the string 'Hello World'.

      *  Flask automatically converts this string into an HTTP response that the browser can display.

+    if `__name__` == '__main__': app.run(debug=True):

     *   This standard Python construct ensures that the server only runs if the script is executed directly (not when it's imported as a module).

      *  app.run(debug=True) starts the built-in development server. Setting debug=True is helpful because it automatically reloads the server when you save changes to the code.
