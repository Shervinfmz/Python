## 🌐 What Is a REST API?

A **REST API (Representational State Transfer)** is a system that lets applications communicate over the internet using standard **HTTP requests**.

REST APIs expose data through **URLs**, and you interact with them using HTTP methods like:

- **GET** → Read data  
- **POST** → Create new data  
- **PUT** → Fully update data  
- **PATCH** → Partially update data  
- **DELETE** → Remove data  

---

## 🔗 Example REST API Endpoint

GitHub API example:

https://api.github.com/users/
<username>


This URL returns public information about a GitHub user.

✔️ You send a **GET request**  
✔️ The server returns **JSON data**

---

## 📥 GET Request Example

### What GET Does

- Retrieves data  
- Does not modify anything  
- Safe  
- Side-effect free  

### Example JSON response:

```json
{
  "login": "shervinfmz",
  "id": 123456,
  "public_repos": 42,
  "followers": 100
}


### 📡 Common HTTP Methods (Summary Table)

| Method     | Description                               |
| ---------- | ----------------------------------------- |
| **GET**    | Retrieve an existing resource (read data) |
| **POST**   | Create a new resource                     |
| **PUT**    | Fully update an existing resource         |
| **PATCH**  | Partially update an existing resource     |
| **DELETE** | Delete a resource                         |



### 🧠 How REST APIs Work (Simple Explanation)

1. Your client (Python code, browser, Postman, etc.) sends a request

2. The server receives it

3. The server processes it

4. The server returns a response — usually JSON

5. Your code parses and uses that data

### 🐍 Python Example — GET Request

````python
import requests

url = "https://api.github.com/users/octocat"
response = requests.get(url)

data = response.json()
print(data["login"])
print(data["public_repos"])
````

## 📘 Python Data Handling Notes

A quick guide covering JSON and basic data-handling techniques.

### 🧩 Working With JSON in Python

Python’s built-in `json` module allows you to easily convert between **Python objects** and **JSON strings**.

### ✔ Convert Python Dictionary → JSON String  
Use **json.dumps()**


````python
import json

data = {"name": "Shervin", "age": 32, "hobby": "bicycle", "coolness": 6}
print(type(data))   # dict

d = json.dumps(data)
print(type(d))      # string (JSON formatted)
````

### ✔ Convert JSON String → Python Dictionary

Use json.loads()

````python
new_data = json.loads(d)
print(new_data)
print(type(new_data))   # dict
````

## 🔎 Summary of JSON functions

| Function               | Input       | Output | Use case                      |
| ---------------------- | ----------- | ------ | ----------------------------- |
| `json.load(file)`      | file object | dict   | Read JSON from a file         |
| `json.loads(string)`   | JSON string | dict   | Parse JSON from an API / text |
| `json.dump(obj, file)` | dict        | file   | Write JSON to a file          |
| `json.dumps(obj)`      | dict        | string | Convert dict to JSON string   |
