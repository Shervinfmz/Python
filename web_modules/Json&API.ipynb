# ## 🧩 Working With JSON in Python

Python’s built-in `json` module allows you to easily convert between **Python objects** and **JSON strings**.

### ✔ Convert Python Dictionary → JSON String  
Use **json.dumps()**


```python
import json

data = {"name": "Shervin", "age": 32, "hobby": "bicycle", "coolness": 6}
print(type(data))   # dict

d = json.dumps(data)
print(type(d))      # string (JSON formatted)
```

### ✔ Convert JSON String → Python Dictionary

Use json.loads()

```python
new_data = json.loads(d)
print(new_data)
print(type(new_data))   # dict
```

## 🔎 Summary of JSON functions

| Function               | Input       | Output | Use case                      |
| ---------------------- | ----------- | ------ | ----------------------------- |
| `json.load(file)`      | file object | dict   | Read JSON from a file         |
| `json.loads(string)`   | JSON string | dict   | Parse JSON from an API / text |
| `json.dump(obj, file)` | dict        | file   | Write JSON to a file          |
| `json.dumps(obj)`      | dict        | string | Convert dict to JSON string   |
