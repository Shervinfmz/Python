# 🛰 Working With a Real API in Python  
Example: How Many People Are in Space Right Now?

In this mini–tutorial we’ll use:

- `requests` → to send an HTTP **GET** request  
- `json` → to work with JSON data  

API used: **Open Notify – People in Space**  
`http://api.open-notify.org/astros.json`

---

````python
import requests
import json
````

### 2️⃣ Send a GET Request to the API

````python
api_url = "http://api.open-notify.org/astros.json"

response = requests.get(api_url)
````

Here:

api_url is the endpoint of the API

requests.get(api_url) sends an HTTP GET request

The server returns a Response object

### 3️⃣ Inspect the Raw Response

````python
print(response)        # <Response [200]> if OK
print(response.text)   # Raw response body as a string
print(response.json)   # ⚠ This prints the *function*, not the data!
````

Notes:

response shows the status, e.g. <Response [200]> → success

response.text is the raw JSON string returned by the server

response.json (without parentheses) is just a reference to the method

To actually get the parsed JSON data, you must call:

````python
json_data = response.json()  # <-- call the function
````

### 4️⃣ Parse JSON Into a Python Dictionary

````python
json_data = response.json()
print(json_data)
````

Now json_data is a Python dictionary, something like:

````python
{
    "message": "success",
    "number": 10,
    "people": [
        {"name": "Sergey Prokopyev", "craft": "ISS"},
        {"name": "Dmitry Petelin", "craft": "ISS"},
        ...
    ]
}
````

You can pretty-print it with the json module:

````python
print(json.dumps(json_data, indent=4))
````

### 5️⃣ Access Specific Fields in the JSON

Because json_data is a dictionary, we can use keys:

````python
print(json_data["number"])   # How many people are in space?
print(json_data["people"])   # List of people (each is a dict)
````

Example to print all names and crafts:

````python
for person in json_data["people"]:
    print(person["name"], "→", person["craft"])
````

### 6️⃣ Full Example Code

````python
import requests
import json

api_url = "http://api.open-notify.org/astros.json"

# Send GET request
response = requests.get(api_url)

# Check basic info
print(response)        # e.g. <Response [200]>
print(response.text)   # raw JSON string
print(response.json)   # function object (for demonstration)

# Parse JSON data
json_data = response.json()
print(json_data)

# Access top-level fields
print("Number of people in space:", json_data["number"])
print("People list:", json_data["people"])

# Nicely formatted output
print("\n👨‍🚀 Astronauts currently in space:")
for person in json_data["people"]:
    print(f"- {person['name']} aboard {person['craft']}")
````

### 7️⃣ Key Takeaways

Use requests.get(url) to send an HTTP GET request.

response.text → raw response string (JSON as text).

response.json() → parses JSON → returns a Python dict.

Don’t forget the parentheses: response.json() not response.json.

Once parsed, you can access JSON fields just like any other dictionary.
